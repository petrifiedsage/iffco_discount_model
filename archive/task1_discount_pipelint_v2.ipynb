```python
import pandas as pd
import numpy as np

df = pd.read_excel("D:\\BlueRose\\iffco_discount_model\\data\\Contract_Data_with_Product Group_IDS_OU_2025.xls")

# Standardize column names
df.columns = df.columns.str.lower()

# Keep only percentage rebate contracts
df = df[df["actual_rebate"].notna()]
df = df[df["actual_rebate"] > 0]

# Drop fixed rebate rows
df = df[df["fixed_rebate_amount"].isna()]

# Parse date
df["created_date"] = pd.to_datetime(
    df["created_date"],
    dayfirst=True,
    errors="coerce"
)

df = df[df["created_date"].notna()]


df["year"] = df["created_date"].dt.year

# Convert START_MONTH
df["start_month"] = pd.to_datetime(
    df["start_month"],
    dayfirst=True,
    errors="coerce"
).dt.month

# Convert END_MONTH
df["end_month"] = pd.to_datetime(
    df["end_month"],
    dayfirst=True,
    errors="coerce"
).dt.month
```

```python
cust_stats = (
    df.groupby("customer")["actual_rebate"]
      .agg(
          customer_avg_rebate="mean",
          customer_contract_count="count"
      )
      .reset_index()
)

df = df.merge(cust_stats, on="customer", how="left")
```

```python
pg_stats = (
    df.groupby("product_group")["actual_rebate"]
      .mean()
      .reset_index(name="product_group_avg_rebate")
)

df = df.merge(pg_stats, on="product_group", how="left")
```

```python
from sklearn.preprocessing import LabelEncoder

encode_cols = [
    "customer",
    "party_site_id",
    "product_group",
    "channel",
    "type",
    "tactic_type",
    "organization",
    "mfg_organization"
]

encoders = {}

for col in encode_cols:
    le = LabelEncoder()
    df[col + "_enc"] = le.fit_transform(df[col].astype(str))
    encoders[col] = le
```

```python
FEATURES = [
    "customer_enc",
    "party_site_id_enc",
    "product_group_enc",
    "channel_enc",
    "type_enc",
    "tactic_type_enc",
    "organization_enc",
    "mfg_organization_enc",
    "year",
    "start_month",
    "end_month",
    "customer_avg_rebate",
    "customer_contract_count",
    "product_group_avg_rebate"
]
```

```python
X = df[FEATURES]
y = df["actual_rebate"]

train_df = df[df["year"] <= 2023]
test_df  = df[df["year"] >= 2024]

X_train = train_df[FEATURES]
y_train = train_df["actual_rebate"]

X_test = test_df[FEATURES]
y_test = test_df["actual_rebate"]

X_train.dtypes
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error

model = RandomForestRegressor(
    n_estimators=300,
    max_depth=12,
    min_samples_leaf=15,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)

preds = model.predict(X_test)

mae = mean_absolute_error(y_test, preds)
print(f"MAE: {mae:.2f}%")
```
